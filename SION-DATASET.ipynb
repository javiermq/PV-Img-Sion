```python
import requests
from pathlib import Path

BASE = "https://dsm.n32x.hevs.ch"
SHARING_ID = "qkgeLrdRT"
PASSWORD = "Sion2026"

def login():
    session = requests.Session()
    session.headers.update({
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
    })

    auth_body = {
        "api": "SYNO.Core.Sharing.Login",
        "method": "login",
        "version": "1",
        "sharing_id": f'"{SHARING_ID}"',
        "password": f'"{PASSWORD}"',
    }

    r = session.post(f"{BASE}/sharing/webapi/auth.cgi", data=auth_body)
    r.raise_for_status()

    data = r.json()
    if not data.get("success"):
        raise RuntimeError(f"Login failed: {data}")

    sharing_sid = data["data"]["sharing_sid"]
    session.cookies.set("sharing_sid", sharing_sid, domain="dsm.n32x.hevs.ch", path="/")

    return session


def list_folder(session, path, limit=1000):
    all_files = []
    offset = 0

    while True:
        body = {
            "api": "SYNO.FolderSharing.List",
            "method": "list",
            "version": "2",
            "folder_path": f'"{path}"',
            "_sharing_id": f'"{SHARING_ID}"',
            "offset": str(offset),
            "limit": str(limit),
            "sort_by": '"name"',
            "sort_direction": '"asc"',
        }

        r = session.post(
            f"{BASE}/sharing/webapi/entry.cgi",
            data=body,
            headers={"Origin": BASE},
        )
        r.raise_for_status()

        data = r.json()
        if not data.get("success"):
            raise RuntimeError(f"List failed for {path}: {data}")

        files = data["data"].get("files", [])
        all_files.extend(files)

        if len(files) < limit:
            break

        offset += limit

    return all_files


def list_month_files(month):
    session = login()

    month_path = f"/June-Aug/{month}"
    days = list_folder(session, month_path)

    all_paths = []

    for day in days:
        if not day.get("isdir"):
            continue

        day_name = day["name"]
        day_path = f"{month_path}/{day_name}"

        print(f"\n📁 {day_path}")

        files = list_folder(session, day_path)

        for f in files:
            name = f["name"]
            remote_path = f"{day_path}/{name}"
            size = f.get("size", "")
            filetype = f.get("type", "")

            print(f"  {remote_path} | size={size} | type={filetype}")
            all_paths.append(remote_path)

    return all_paths
```

```python
MONTH = "06"

if __name__ == "__main__":
    files = list_month_files(MONTH)

    print("\n" + "=" * 60)
    print(f"Total archivos en mes {MONTH}: {len(files)}")

    # Opcional: guardar listado en TXT
    out = Path(f"files_month_{MONTH}.txt")
    out.write_text("\n".join(files), encoding="utf-8")
    print(f"Listado guardado en: {out}")
```
